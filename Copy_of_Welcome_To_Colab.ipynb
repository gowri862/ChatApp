```
import numpy as np
# Define structured array data type
dtype1 = [
	('name', 'S10'),	# Fixed-length string (10 bytes)
	('height', float),  # Height as float
	('age', int),   	# Age as integer
	('gender', 'S10')   # Gender as string
]
# Data values
values = [
	('Shubham', 5.9, 23, 'M'),
	('Arpita', 5.6, 23, 'F'),
	('Vaishali', 5.2, 30, 'F')
]
# Create structured array
x = np.array(values, dtype=dtype1)
print("Original array:")
print(x)

# Sort by age
y = np.sort(x, order='age')
print("\nSorted by age:")
print(y)
# Sort by age, then height
z = np.sort(x, order=['age', 'height'])
print("\nSorted by age, then height:")
print(z)

```
