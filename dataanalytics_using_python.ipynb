```
import numpy as np
# Define structured array data type
dtype1 = [
	('name', 'S10'),	# Fixed-length string (10 bytes)
	('height', float),  # Height as float
	('age', int),   	# Age as integer
	('gender', 'S10')   # Gender as string
]
# Data values
values = [
	('Shubham', 5.9, 23, 'M'),
	('Arpita', 5.6, 23, 'F'),
	('Vaishali', 5.2, 30, 'F')
]
# Create structured array
x = np.array(values, dtype=dtype1)
print("Original array:")
print(x)

# Sort by age
y = np.sort(x, order='age')
print("\nSorted by age:")
print(y)
# Sort by age, then height
z = np.sort(x, order=['age', 'height'])
print("\nSorted by age, then height:")
print(z)

```

```
import numpy as np

#array
dirty_temps = np.array([22.5, np.nan, 100.2, 19.8, -85.0, 25.1, np.nan, 21.0])

#Mark sensor errors as NaN
dirty_temps[(dirty_temps < -50) | (dirty_temps > 60)] = np.nan

#Calculate average of valid temperatures
avg_temp = np.nanmean(dirty_temps)

#Replace NaN values with the average
dirty_temps[np.isnan(dirty_temps)] = avg_temp

#Normalize using Min-Max Scaling
min_temp = np.min(dirty_temps)
max_temp = np.max(dirty_temps)

normalized_temps = (dirty_temps - min_temp) / (max_temp - min_temp)

print("Average Temperature:", avg_temp)
print("Cleaned Temperatures:", dirty_temps)
print("Normalized Temperatures:", normalized_temps)
```

```
import pandas as pd
# 1. Load the dataset
df = pd.read_csv("/content/student_details - student_details.csv")
# print("Loaded Data:")
# print(df)
# print("\n" + "-"*50 + "\n")
# # 2. Basic information about the data
# print("Dataset Info:")
# print(df.info())
# print("\n" + "-"*50 + "\n")
# print("Statistical Summary:")
# print(df.describe())
# print("\n" + "-"*50 + "\n")
# # 3. Check for missing values
# print("Missing Values:")
# print(df.isna())
# print("\nTotal Missing Values Per Column:")
# print(df.isna().sum())
# print("\n" + "-"*50 + "\n")
 # 4. Fill missing 'Marks' with mean
Subject_mean=['Physics','Maths','Chemistry','English','Computer_Science']
T_Mean=df[Subject_mean].mean()
print("All Subject Mean")
print(T_Mean)
print("--------")
df['Total'] = df.iloc[:, 3:8].sum(axis=1)
print("Toatl Mark")
print(df[['Name', 'Total']])
print("--------")
#Average
df['Average'] =df['Total']/5
print("Average Mark per Subject")
print(df)
print("--------")
#Average>80
high_scorers = df[df['Average'] > 80]
print("Students with Marks > 80%:")
print(high_scorers)
print("\n" + "-"*50 + "\n")
#Add New Cloumn
df['Grade'] = ['A' if avg >= 85 else 'B' for avg in df['Average']]
print("Added 'Grade' Column:")
print(df)
print("\n" + "-"*50 + "\n")
#Group
Dept_avg = df.groupby('Department')['Total'].mean()
print(Dept_avg)
#save the file
df.to_csv("student_updated.csv", index=False)

```

```
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
iris_data = pd.read_csv("/content/iris - iris.csv")
print(iris_data.head())
print(iris_data.describe())
grouped_data = iris_data.groupby('variety').describe()
plt.figure(figsize=(10, 5))
plt.subplot(1, 2, 1)
plt.hist(iris_data['sepal.length'], bins=20, color='blue')
plt.title('Sepal Length Distribution')
plt.xlabel('Sepal Length')
plt.ylabel('Frequency')
plt.subplot(1, 2, 2)
plt.hist(iris_data['petal.width'], bins=20, color='orange')
plt.title('Petal Width Distribution')
plt.xlabel('Petal Width')
plt.ylabel('Frequency')
plt.tight_layout()
plt.show()
```

```
import pandas as pd
import matplotlib.pyplot as plt
file = "sales_data.xlsx"
sales_df=pd.read_excel(file, sheet_name='Sales')
emp_df=pd.read_excel(file, sheet_name='Employees')
print(sales_df)
print(emp_df)

#Fill Missing Data
sales_df["Quantity"] = sales_df["Quantity"].fillna(sales_df["Quantity"].median())
sales_df["Price"] = sales_df["Price"].fillna(sales_df["Price"].mean())
print(sales_df)

# Total Amount Column
sales_df["Total Amount"] = sales_df["Quantity"] * sales_df["Price"]
print(sales_df)

#Calculate Total Revenue
total_revenue = sales_df["Total Amount"].sum()
print(f"Total Revenue: {total_revenue}")

#Calculate Average Revenue
average_revenue = sales_df["Total Amount"].mean()
print(f"Average Revenue: {average_revenue}")

#Calculate Product Wise Revenue
product_revenue = sales_df.groupby("Product")["Total Amount"].sum()
print("\nProduct Wise Revenue:")
print(product_revenue)

#Calculate city Wise Revenue
city_revenue = sales_df.groupby("City")["Total Amount"].sum()
print("\nCity Wise Revenue:")
print(city_revenue)

# #Bar Chart
# plt.figure(figsize=(5, 3))
# plt.bar(product_revenue.index, product_revenue.values)
# plt.xlabel("Product")
# plt.ylabel("Total Revenue")
# plt.title("Product-wise Revenue")
# plt.show()
# #city-wise
# plt.figure(figsize=(5, 6))
# city_revenue.plot(kind='line', color='pink')
# plt.xlabel("City")
# plt.ylabel("Total Revenue")
# plt.title("City-wise Revenue")
# plt.show()

#Calculate Average salary
average_salary = emp_df["Salary"].mean()
print(f"Average Salary: {average_salary}")

#Calculate Highest Salary
highest_salary = emp_df["Salary"].max()
print(f"Highest Salary: {highest_salary}")

#Calculate Employees with Salary > 50000
high_salary_employees = emp_df[emp_df["Salary"] > 50000]
print("\nEmployees with Salary > 50000:")
print(high_salary_employees)

#Calculate Employees with Salary > 50000
high_salary_employees = emp_df[emp_df["Salary"] > 50000]
print("\nEmployees with Salary > 50000:")
print(high_salary_employees)

#Calculate Total Number of Employees
total_employees = emp_df["Name"].count()
print(f"Total Number of Employees: {total_employees}")

#Calculate Total Number of Orders
total_orders = sales_df["OrderID"].count()
print(f"Total Number of Orders: {total_orders}")


#Calculate Total Number of Quantity
total_orders = sales_df["Quantity"].sum()
print(f"Total Orders Quantity: {total_orders}")

#Draw a bar chart with Department wise Average Salary
department_salary = emp_df.groupby("Dept")["Salary"].mean()
plt.figure(figsize=(5, 3))
department_salary.plot(kind='pie', color='green')
plt.xlabel("Department")
plt.ylabel("Average Salary")
plt.title("Department-wise Average Salary")
plt.show()


```
